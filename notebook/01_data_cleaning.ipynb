```python
import pandas as pd

raw_path = "../data/raw/nifty_50_yfinance_raw.csv"

df = pd.read_csv(
    raw_path,
    skiprows=2,
    parse_dates=['Date']
)

df.head()
```

```python
df.columns = [
    "Date",
    "Close",
    "High",
    "Low",
    "Open",
    "Volume"
]

df.head()
```

```python
df = df.sort_values("Date")
df = df.set_index("Date")

df.info()
```

```python
df = df[["Open", "High", "Low", "Close", "Volume"]]
df.head()
```
